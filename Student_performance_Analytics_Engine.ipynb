```python
import numpy as np
```

Import,seed, generating  500*6 marks matrix


```python
np.random.seed(7)
students=500
subjects=6
marks=np.random.normal(65,12,(students,subjects))
marks=np.clip(marks,0,100)
marks=np.round(marks,2)
print("Marks Matrix:")
print(marks[:5])
print("Shape of Matrix:",marks.shape)
```

Sum mean median std var with axis=0 and axis=1


```python
print("Subject wise")
print("Sum:",np.sum(marks,axis=0))
print("Mean:",np.mean(marks,axis=0))
print("Median:",np.median(marks,axis=0))
print("Standard Deviation:",np.std(marks,axis=0))
print("Variance:",np.var(marks,axis=0))
print("Student wise")
print("sum:",np.sum(marks,axis=1)[:5])
print("mean:",np.mean(marks,axis=1)[:5])
print("median ",np.median(marks,axis=1)[:5])
print("standard deviation ",np.std(marks,axis=1)[:5])
print("Variance ",np.var(marks, axis=1)[:5])
```

Z-scores via broadcasting; flag outliers; rank students

```python
mean=np.mean(marks,axis=0)
std=np.std(marks,axis=0)
z=(marks-mean)/std
outliers=np.abs(z)>2
print("Number of Outliers:",np.sum(outliers))
total=np.sum(marks,axis=1)
rank=np.argsort(-total)
print("\nTop 10 Students")
for i in range(10):
      student = rank[i]
print(i+1)
print(student+1)
print(round(total[student], 2))
```

Add totals with hstack; export CSV with files.download

```python
total=np.sum(marks,axis=1)
average=np.mean(marks,axis=1)
total=total.reshape(500,1)
average=average.reshape(500,1)
final_marks=np.hstack((marks,total,average))
print("Final Matrix Shape:",final_marks.shape)
print(final_marks[:5])
np.savetxt(
    "student_performance.csv",
    final_marks,
    delimiter=",",
    fmt="%.2f",
    header="Maths,Physics,Chemistry,Computer,English,Statistics,Total,Average",
    comments=""
)
print("CSV file saved successfully!")
```
